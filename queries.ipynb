# Prepping Data for Querrying

```python
import pandas as pd

sales = pd.read_json("data/verified_sales.json")

sales["value"] = sales["price"] * sales["quantity"]
```

# Queries

## Total Revenue:

```python
print("€" + (str)(sales["value"].sum()))
```

## Sales by Category:


```python
sales.groupby("category")["value"].sum().sort_values(ascending=False)
```

## Top Customer:

```python
sales.groupby("customer")["value"].sum().sort_values(ascending=False).head(1)
```

## Top Products:

```python
max_quantity = sales.groupby("product")["quantity"].sum()
max_quantity[max_quantity == max_quantity.max()]
```

## Most Valueable Day:

```python
sales.groupby("date")["value"].sum().sort_values(ascending=False).head(1)
```
