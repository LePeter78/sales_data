# Import

```python
import pandas as pd
```

# Read Data

```python
sales_data = pd.read_csv("data/raw_data.csv")
sales_data
```

# Filter Dirty Data

```python
sales_data["date"] = pd.to_datetime(sales_data["date"], errors="coerce")

sales_data["price"] = pd.to_numeric(sales_data["price"], errors="coerce")
sales_data.loc[sales_data["price"] < 0, "price"] = pd.NA

sales_data["quantity"] = pd.to_numeric(sales_data["quantity"], errors="coerce")
sales_data.loc[sales_data["quantity"] < 0, "quantity"] = pd.NA

print(sales_data.isna().sum())
sales_data[sales_data.isna().any(axis=1)]
```

# Write Data

```python
sales_data[sales_data.isna().any(axis=1)].to_json("data/faulty_data.json", indent=4, date_format="iso")
sales_data.dropna().to_json("data/verified_sales.json", indent=4, date_format="iso")
sales_data.to_json("data/all_sales.json", indent=4, date_format="iso")
```
